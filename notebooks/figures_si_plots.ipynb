```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ------------------------------------------------------------------
# Dataset
# ------------------------------------------------------------------

df = pd.DataFrame(
    main_text_figure_data,
    columns=["Paper","Total","Relevant","Novel"]
)

# ------------------------------------------------------------------
# Aggregate counts
# ------------------------------------------------------------------

total = df["Total"].sum()
relevant = df["Relevant"].sum()
novel = df["Novel"].sum()

not_relevant = total - relevant
redundant = relevant - novel

print(total, relevant, novel)

# ------------------------------------------------------------------
# Figure
# ------------------------------------------------------------------

fig = make_subplots(
    rows=1,
    cols=2,
    specs=[[{"type":"funnel"},{"type":"pie"}]],
    column_widths=[0.55,0.45],
    subplot_titles=[
        "Information Reduction Pipeline",
        "Composition of All Figures"
    ]
)

# Funnel
fig.add_trace(
    go.Funnel(
        y=[
            "All figures",
            "Schema-relevant",
            "Novel information"
        ],
        x=[total, relevant, novel],
        textinfo="value+percent previous",
        marker=dict(color=["#4C78A8","#F58518","#54A24B"])
    ),
    row=1,
    col=1
)

# Pie
fig.add_trace(
    go.Pie(
        labels=[
            "Not relevant",
            "Relevant but\nalready in text",
            "Novel"
        ],
        values=[
            not_relevant,
            redundant,
            novel
        ],
        hole=0.45,
        textinfo="percent+label",
        marker=dict(colors=[
            "#BDBDBD",
            "#F58518",
            "#54A24B"
        ])
    ),
    row=1,
    col=2
)

fig.update_layout(
    template="plotly_white",
    height=600,
    width=1100,
    title=dict(
        text="Utility of Figures for Information Extraction",
        x=0.5
    ),
    font=dict(size=15),
    margin=dict(t=80,b=80,l=40,r=40)
)
fig.update_layout(showlegend=False)

fig.show()
```
